# 03 — Model improvements and preprocessing variants

The source code provides separate experiment names: `class_weighted`, `finetuned`, and `letterbox`. The fine-tuned workflow uses class weights and unfreezes VGG16 block 5 in a second, lower-learning-rate stage. The letterbox workflow also uses class weighting and block-5 fine-tuning, so its historical validation result does not isolate letterboxing. Every run gets a new timestamped output directory.

## Setup

Set paths and inspect the selected experiment before training. No experiment is started automatically.

```python
from pathlib import Path
import os
import subprocess
import sys

# In Colab, set this to the path where you cloned/uploaded the repository.
PROJECT_ROOT = Path(os.environ.get("PROJECT_ROOT", "/content/Breast-Cancer-Detection-VGG16"))
if not (PROJECT_ROOT / "src" / "train.py").is_file():
    raise FileNotFoundError(f"Set PROJECT_ROOT to the repository folder. Current value: {PROJECT_ROOT}")
sys.path.insert(0, str(PROJECT_ROOT))

# Optional Drive integration. Leave False to use Colab runtime storage.
USE_GOOGLE_DRIVE = False
if USE_GOOGLE_DRIVE:
    from google.colab import drive
    drive.mount("/content/drive")

_DEFAULT_DATA = "/content/drive/MyDrive/BreastCancerProject/BUSI" if USE_GOOGLE_DRIVE else "/content/data/BUSI"
_DEFAULT_OUTPUT = "/content/drive/MyDrive/BreastCancerProject/reproducible_runs" if USE_GOOGLE_DRIVE else "/content/busi_runs"
BUSI_ROOT = Path(os.environ.get("BUSI_ROOT", _DEFAULT_DATA))
OUTPUT_ROOT = Path(os.environ.get("PROJECT_OUTPUT_ROOT", _DEFAULT_OUTPUT))
RUN_TRAINING = False  # Change to True only when you intend to train this experiment.
```

## Choose one experiment

Set `EXPERIMENT` to one value and then explicitly enable the run.

```python
EXPERIMENT = "class_weighted"  # choose: class_weighted, finetuned, letterbox
if EXPERIMENT not in {"class_weighted", "finetuned", "letterbox"}:
    raise ValueError("Choose a supported improvement experiment")
print("Selected experiment:", EXPERIMENT)
```

```python
if RUN_TRAINING:
    command = [sys.executable, "-m", "src.train", "--experiment", EXPERIMENT,
               "--data-root", str(BUSI_ROOT), "--output-root", str(OUTPUT_ROOT)]
    subprocess.run(command, cwd=PROJECT_ROOT, check=True)
else:
    print("Training is disabled. Set RUN_TRAINING = True after reviewing the experiment.")
```
